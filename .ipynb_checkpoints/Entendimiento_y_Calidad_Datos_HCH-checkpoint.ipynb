```python
# ── CELDA 0 — Setup global ───────────────────────────────────────
import sqlite3
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from datetime import datetime

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 140)

DB_PATH = "mercado_hch.db"   # ajusta si tu archivo está en otra carpeta

def query(sql, params=None):
    with sqlite3.connect(DB_PATH) as con:
        return pd.read_sql(sql, con, params=params)

print(f"✅ Conectado a: {DB_PATH}")
print(f"   Fecha de corrida del notebook: {datetime.now().strftime('%Y-%m-%d %H:%M')}")
```

# 📦 Entregable 1 — Perfilamiento

## 1.1 Entendimiento (por tabla)

Para cada tabla: volumen, esquema (`PRAGMA table_info`), muestra de registros y qué representa una fila.

```python
# ── 1.1.a — Inventario de tablas y volumen ───────────────────────
tablas = query("SELECT name FROM sqlite_master WHERE type='table' AND name NOT LIKE 'sqlite_%'")['name'].tolist()

resumen_volumen = []
for t in tablas:
    n = query(f"SELECT COUNT(*) AS n FROM {t}")['n'].iloc[0]
    resumen_volumen.append({'tabla': t, 'filas': n})

df_volumen = pd.DataFrame(resumen_volumen).sort_values('filas', ascending=False)
print("📊 Volumen por tabla:")
display(df_volumen)
```

```python
# ── 1.1.b — printSchema de cada tabla ─────────────────────────────
for t in tablas:
    print(f"\n{'='*60}\n📋 TABLA: {t}\n{'='*60}")
    esquema = query(f"PRAGMA table_info({t})")
    display(esquema[['name', 'type', 'notnull', 'pk']])
```

```python
# ── 1.1.c — Muestra de registros por tabla ────────────────────────
muestras = {}
for t in tablas:
    muestras[t] = query(f"SELECT * FROM {t} ORDER BY ROWID DESC LIMIT 5")
    print(f"\n--- Muestra de '{t}' (últimos 5 registros insertados) ---")
    display(muestras[t])
```

## 1.2 Reglas de negocio

Validamos las reglas específicas que ya identificamos para `bmc_precios` (la tabla más sensible, porque viene de OCR).

```python
# ── Regla 1 — Consistencia interna: precio_cop_kg ≈ total_cop / cantidad_kg ──
# Esta es la CAPA 1 de verificación del OCR: si el regex/parser leyó bien los 3
# números de la línea del boletín, esta igualdad debe cumplirse casi exactamente.

df_bmc = query("""
    SELECT fecha, n_rueda, producto, producto_bmc, archivo_pdf,
           cantidad_kg, total_cop, precio_cop_kg, precio_usd_ton, trm
    FROM bmc_precios
    WHERE cantidad_kg IS NOT NULL AND cantidad_kg != 0
      AND total_cop IS NOT NULL AND precio_cop_kg IS NOT NULL
""")

df_bmc['precio_calculado'] = df_bmc['total_cop'] / df_bmc['cantidad_kg']
df_bmc['error_relativo_pct'] = (
    (df_bmc['precio_cop_kg'] - df_bmc['precio_calculado']).abs()
    / df_bmc['precio_calculado'].replace(0, np.nan)
) * 100

UMBRAL_SOSPECHA = 5  # % de error relativo
df_bmc['sospechoso_regla1'] = df_bmc['error_relativo_pct'] > UMBRAL_SOSPECHA

n_sospechosos = df_bmc['sospechoso_regla1'].sum()
pct_sospechosos = n_sospechosos / len(df_bmc) * 100

print(f"Regla 1 — precio_cop_kg vs total_cop/cantidad_kg")
print(f"  Filas evaluadas:    {len(df_bmc):,}")
print(f"  Filas sospechosas:  {n_sospechosos:,} ({pct_sospechosos:.1f}%)  [error relativo > {UMBRAL_SOSPECHA}%]")
print(f"\nTop 15 filas más sospechosas (mayor error relativo):")
display(
    df_bmc.sort_values('error_relativo_pct', ascending=False)
    [['fecha','producto','cantidad_kg','total_cop','precio_cop_kg','precio_calculado','error_relativo_pct','archivo_pdf']]
    .head(15)
)

print(f"\nSospechosos por producto (Regla 1):")
display(
    df_bmc.groupby('producto')['sospechoso_regla1']
    .agg(filas='count', sospechosas='sum')
    .assign(pct=lambda d: (d['sospechosas']/d['filas']*100).round(1))
    .sort_values('pct', ascending=False)
)
```

```python
#Paso 1 — Identificar filas a corregir (con rowid, para precisión exacta)
import sqlite3
import pandas as pd
import numpy as np

DB_PATH = "mercado_hch.db"

with sqlite3.connect(DB_PATH) as con:
    df = pd.read_sql("""
        SELECT fecha, n_rueda, producto_bmc, producto, archivo_pdf,
               cantidad_kg, total_cop, precio_cop_kg, precio_usd_ton, trm
        FROM bmc_precios
        WHERE cantidad_kg IS NOT NULL AND cantidad_kg != 0
          AND total_cop IS NOT NULL AND precio_cop_kg IS NOT NULL
    """, con)

# Verificación rápida de que la columna existe antes de seguir
print("Columnas disponibles:", df.columns.tolist())

df['precio_calculado'] = df['total_cop'] / df['cantidad_kg']
df['error_relativo_pct'] = (
    (df['precio_cop_kg'] - df['precio_calculado']).abs()
    / df['precio_calculado'].replace(0, np.nan)
) * 100

sospechosos = df[df['error_relativo_pct'] > 5].copy()
print(f"Total sospechosos: {len(sospechosos)}")

EXCLUIR = [
    ('2024-03-20', 'subproductos_carnicos'),
    ('2023-03-14', 'sebo_res'),
]
mask_excluir = sospechosos.apply(lambda r: (r['fecha'], r['producto']) in EXCLUIR, axis=1)
a_corregir = sospechosos[~mask_excluir].copy()
pendientes_manual = sospechosos[mask_excluir].copy()

print(f"A corregir automáticamente: {len(a_corregir)}")
print(f"Pendientes de revisión manual (excluidas): {len(pendientes_manual)}")
```

```python
# Paso 2 Log auditoría con clave natural
a_corregir['precio_cop_kg_nuevo'] = a_corregir['precio_calculado'].round(2)
a_corregir['precio_usd_ton_nuevo'] = (
    (a_corregir['precio_cop_kg_nuevo'] / a_corregir['trm']) * 1000
).round(2)

log_correccion = a_corregir[[
    'fecha', 'n_rueda', 'producto_bmc', 'producto', 'archivo_pdf',
    'precio_cop_kg', 'precio_cop_kg_nuevo',
    'precio_usd_ton', 'precio_usd_ton_nuevo',
    'error_relativo_pct'
]].rename(columns={
    'precio_cop_kg': 'precio_cop_kg_original',
    'precio_usd_ton': 'precio_usd_ton_original'
})
log_correccion['motivo'] = 'Regla 1 - confirmado contra PDF original (BDR_197, BDR_126): precio_cop_kg danado por OCR, total_cop/cantidad_kg correcto'
log_correccion['fecha_correccion'] = pd.Timestamp.now().strftime('%Y-%m-%d %H:%M')

with sqlite3.connect(DB_PATH) as con:
    log_correccion.to_sql('bmc_precios_correcciones_log', con, if_exists='append', index=False)

print(f"✅ Log guardado: {len(log_correccion)} filas")
display(log_correccion[['fecha','producto','precio_cop_kg_original','precio_cop_kg_nuevo']])
```

```python
# Paso 3 (corregido) — UPDATE por clave natural, con verificación de unicidad antes de tocar nada
with sqlite3.connect(DB_PATH) as con:
    cur = con.cursor()
    aplicadas = 0
    for _, row in a_corregir.iterrows():
        # Verifica que la clave identifica exactamente 1 fila antes de actualizar
        cur.execute("""
            SELECT COUNT(*) FROM bmc_precios
            WHERE fecha = ? AND n_rueda = ? AND producto_bmc = ?
        """, (row['fecha'], row['n_rueda'], row['producto_bmc']))
        n_match = cur.fetchone()[0]
        if n_match != 1:
            print(f"⚠️  Saltada (match={n_match}, esperado 1): {row['fecha']} / {row['producto_bmc']}")
            continue

        cur.execute("""
            UPDATE bmc_precios
            SET precio_cop_kg = ?, precio_usd_ton = ?
            WHERE fecha = ? AND n_rueda = ? AND producto_bmc = ?
        """, (row['precio_cop_kg_nuevo'], row['precio_usd_ton_nuevo'],
              row['fecha'], row['n_rueda'], row['producto_bmc']))
        aplicadas += 1
    con.commit()

print(f"✅ {aplicadas} de {len(a_corregir)} filas corregidas en bmc_precios")
```

```python
import sqlite3
import pandas as pd
import re

DB_PATH = "mercado_hch.db"

with sqlite3.connect(DB_PATH) as con:
    archivos_simm = pd.read_sql(
        "SELECT DISTINCT archivo_pdf FROM bmc_precios WHERE archivo_pdf LIKE '%SIMM%'", con
    )['archivo_pdf'].tolist()

# Regex tolerante a espacio antes del guion: "N -134" o "N-134"
patron = re.compile(r'N\s*-\s*(\d+)')

actualizaciones = []
for archivo in archivos_simm:
    m = patron.search(archivo.upper())
    if m:
        n_rueda_correcto = int(m.group(1))
        actualizaciones.append((archivo, n_rueda_correcto))
    else:
        print(f"⚠️  No se pudo extraer n_rueda de: {archivo}")

print("n_rueda extraído por archivo:")
for archivo, n in actualizaciones:
    print(f"  {n:>4}  ←  {archivo}")
```

```python
with sqlite3.connect(DB_PATH) as con:
    cur = con.cursor()
    total = 0
    for archivo, n_rueda_correcto in actualizaciones:
        cur.execute(
            "UPDATE bmc_precios SET n_rueda = ? WHERE archivo_pdf = ? AND n_rueda IS NULL",
            (n_rueda_correcto, archivo)
        )
        total += cur.rowcount
    con.commit()
print(f"✅ {total} filas actualizadas con n_rueda corregido")
```

**Fecha de la sesión:** completar con fecha real
**Tabla auditada:** `bmc_precios` (9,073 filas evaluadas)

### Resultado final
- **Sospechosos iniciales:** 23 filas (0.3%)
- **Sospechosos finales:** 0 filas
- **Correcciones documentadas en `bmc_precios_correcciones_log`:** 23

### Hallazgos y causas raíz

| Hallazgo | Causa raíz | Filas | Cómo se corrigió |
|---|---|---|---|
| Precio con patrón ×10 / dígito sustituido | OCR dañó específicamente la columna "Precio promedio" del boletín (cantidad_kg y total_cop sí eran confiables) | 19 | `precio_cop_kg = total_cop / cantidad_kg`, recalculando `precio_usd_ton` |
| `n_rueda = NULL` en boletines "SIMM" | Regex `extraer_fecha_y_rueda()` esperaba `N-134` pegado, pero el nombre del archivo tenía `N -134` (espacio antes del guion) | 31 (1 de ellas coincidía también con el patrón de precio) | Regex tolerante a espacio: `r'N\s*-\s*(\d+)'`, extrayendo el número directo del nombre del archivo |
| `BDR_100` sin ninguna fila | Es un boletín de "Cesión de Derechos Económicos", sin tabla de producto — comportamiento esperado, no un error | 0 | Ninguna — se confirma que no es un hueco de datos |
| `sebo_res` 2023-03-14 — los 3 campos dañados (cantidad_kg, total_cop, precio_cop_kg) | Probable desincronización del regex con un número de 4 grupos de miles (`1,177,688.89`) — quedó solo el segundo grupo (`177`) | 1 | Reemplazo directo con los 3 valores reales verificados contra el PDF (`BDR_51`, Rueda 51) |
| `subproductos_carnicos` 2024-03-20 — `total_cop` truncado | A `total_cop` le faltaban los últimos 3 dígitos (`4,947` en vez de `4,947,440`) | 1 | `total_cop` corregido con el valor real del PDF (`BDR_56`, Rueda 56), luego `precio_cop_kg` recalculado |

### Método de verificación
Cada corrección de magnitud (no solo las de NULL/regex) se confirmó **contra el PDF original**, no solo contra el cálculo derivado — abrimos `BDR_197` (harina_hueso), `BDR_126` (HCH), `BDR_51` (sebo_res) y `BDR_56` (subproductos_carnicos) y comparamos línea por línea.

```python
import sqlite3
import pandas as pd
import numpy as np

DB_PATH = "mercado_hch.db"

with sqlite3.connect(DB_PATH) as con:
    df_bmc = pd.read_sql("SELECT * FROM bmc_precios WHERE precio_cop_kg IS NOT NULL", con)

# Límites IQR por producto (mismo método que usamos en Validez)
limites = {}
for prod, grupo in df_bmc.groupby('producto'):
    precios = grupo['precio_cop_kg'].dropna()
    if len(precios) < 5:
        continue
    Q1, Q3 = precios.quantile(0.25), precios.quantile(0.75)
    IQR = Q3 - Q1
    limites[prod] = (max(Q1 - 1.5*IQR, 0), Q3 + 1.5*IQR)

mask = df_bmc.apply(
    lambda r: r['producto'] in limites and limites[r['producto']][0] <= r['precio_cop_kg'] <= limites[r['producto']][1],
    axis=1
)
df_bmc_limpio = df_bmc[mask].copy()

print(f"bmc_precios:        {len(df_bmc):,} filas")
print(f"bmc_precios_limpio: {len(df_bmc_limpio):,} filas ({len(df_bmc)-len(df_bmc_limpio)} excluidas por outlier de mercado)")

with sqlite3.connect(DB_PATH) as con:
    df_bmc_limpio.to_sql('bmc_precios_limpio', con, if_exists='replace', index=False)
print("✅ bmc_precios_limpio regenerada")
```

```python
# ── Regla 2 — n_rueda debe ser razonablemente secuencial en el tiempo ─────
df_ruedas = query("""
    SELECT DISTINCT fecha, n_rueda FROM bmc_precios
    WHERE n_rueda IS NOT NULL
    ORDER BY fecha
""")
df_ruedas['fecha'] = pd.to_datetime(df_ruedas['fecha'])

# Dentro de cada año, n_rueda debería crecer de forma aproximadamente monótona
df_ruedas['anio'] = df_ruedas['fecha'].dt.year
df_ruedas['diff_rueda'] = df_ruedas.groupby('anio')['n_rueda'].diff()

anomalias_rueda = df_ruedas[(df_ruedas['diff_rueda'] < -5) | (df_ruedas['diff_rueda'] > 30)]
print(f"Regla 2 — secuencialidad de n_rueda por año")
print(f"  Anomalías encontradas: {len(anomalias_rueda)} (saltos o retrocesos grandes en n_rueda)")
if len(anomalias_rueda) > 0:
    display(anomalias_rueda.head(15))
```

```python
# ── Regla 3 — Consistencia: precio_usd_ton vs (precio_cop_kg / trm) * 1000 ──
df_trm = query("""
    SELECT fecha, producto, precio_cop_kg, precio_usd_ton, trm, archivo_pdf
    FROM bmc_precios
    WHERE precio_cop_kg IS NOT NULL AND precio_usd_ton IS NOT NULL
      AND trm IS NOT NULL AND trm != 0
""")
df_trm['usd_calculado'] = (df_trm['precio_cop_kg'] / df_trm['trm']) * 1000
df_trm['error_relativo_pct'] = (
    (df_trm['precio_usd_ton'] - df_trm['usd_calculado']).abs()
    / df_trm['usd_calculado'].replace(0, np.nan)
) * 100

n_sosp_regla3 = (df_trm['error_relativo_pct'] > 1).sum()  # tolerancia de redondeo, 1%
print(f"Regla 3 — precio_usd_ton vs precio_cop_kg/trm*1000")
print(f"  Filas evaluadas:   {len(df_trm):,}")
print(f"  Inconsistentes:    {n_sosp_regla3:,} ({n_sosp_regla3/len(df_trm)*100:.1f}%)")
```

```python
# ── Regla 4 — Fechas dentro de rango razonable (no futuras, no anteriores a 2015) ─
hoy = pd.Timestamp.today()
for t, col in [('bmc_precios', 'fecha'), ('sacrificio_bovino', 'fecha'), ('hch_precio_unificado', 'fecha')]:
    try:
        df_f = query(f"SELECT {col} FROM {t} WHERE {col} IS NOT NULL")
        df_f[col] = pd.to_datetime(df_f[col], errors='coerce')
        futuras = (df_f[col] > hoy).sum()
        antiguas = (df_f[col] < pd.Timestamp('2015-01-01')).sum()
        invalidas = df_f[col].isna().sum()
        print(f"{t}.{col}: {futuras} fechas futuras | {antiguas} antes de 2015 | {invalidas} no parseables")
    except Exception as e:
        print(f"{t}: no se pudo evaluar ({e})")
```

**Resumen de reglas de negocio violadas** (completar con los números reales de arriba):

| Regla | Qué valida | % de filas que la violan |
|---|---|---|
| 1. `precio_cop_kg ≈ total_cop/cantidad_kg` | Que el OCR/parser leyó los 3 números en el orden correcto | ✅ |
| 2. `n_rueda` secuencial | Que la fecha/rueda extraída del nombre del PDF tiene sentido | ✅ |
| 3. `precio_usd_ton ≈ precio_cop_kg/trm*1000` | Que la conversión de moneda fue correcta | ✅ |
| 4. Fechas dentro de rango razonable | Que no hay errores de parseo de fecha | ✅ |

```python
# ── UNICIDAD — duplicados ──────────────────────────────────────────────────
dup_bmc = query("""
    SELECT fecha, n_rueda, producto_bmc, COUNT(*) AS n
    FROM bmc_precios
    GROUP BY fecha, n_rueda, producto_bmc
    HAVING COUNT(*) > 1
""")
print(f"Duplicados en bmc_precios (mismo fecha+n_rueda+producto_bmc): {len(dup_bmc)}")
if len(dup_bmc) > 0:
    display(dup_bmc.head(10))

try:
    dup_imp = query("""
        SELECT producto, fecha, COUNT(*) AS n
        FROM importaciones_limpias
        GROUP BY producto, fecha
        HAVING COUNT(*) > 1
    """)
    print(f"\nDuplicados en importaciones_limpias (mismo producto+fecha): {len(dup_imp)}")
    if len(dup_imp) > 0:
        display(dup_imp.head(10))
except Exception as e:
    print(f"importaciones_limpias: {e}")
```

```python
# ── VALIDEZ — outliers por producto (IQR + magnitud extrema) ──────────────
# En vez de un rango global hardcoded (ej. BETWEEN 50 AND 5000 para TODO),
# calculamos límites IQR específicos por producto, y además marcamos
# "extremos" (posible bug de formato numérico, no solo outlier de mercado).

resultados_validez = []
for prod, grupo in df_bmc_full.groupby('producto'):
    precios = grupo['precio_cop_kg'].dropna()
    if len(precios) < 5:
        continue
    Q1, Q3 = precios.quantile(0.25), precios.quantile(0.75)
    IQR = Q3 - Q1
    lim_inf, lim_sup = Q1 - 1.5*IQR, Q3 + 1.5*IQR
    outliers_iqr = ((precios < lim_inf) | (precios > lim_sup)).sum()

    mediana = precios.median()
    # "Extremo": más de 50x la mediana -> casi seguro error de magnitud (no outlier de mercado)
    extremos = (precios > mediana * 50).sum() if mediana > 0 else 0

    resultados_validez.append({
        'producto': prod,
        'n': len(precios),
        'mediana_cop_kg': round(mediana, 0),
        'lim_sup_iqr': round(lim_sup, 0),
        'outliers_iqr': outliers_iqr,
        'pct_outliers_iqr': round(outliers_iqr/len(precios)*100, 1),
        'extremos_50x_mediana': extremos,
        'pct_extremos': round(extremos/len(precios)*100, 1),
    })

df_validez = pd.DataFrame(resultados_validez).sort_values('pct_extremos', ascending=False)
print("📐 Validez — outliers IQR vs valores extremos (posible bug de magnitud):")
display(df_validez)
```

# 🔧 ¿Hay que volver a correr el OCR?

**Contexto del bug (recordatorio):** la función `a_float()` original del scraper (`BMC_Scraper_Final`) solo
eliminaba comas (`s.replace(',', '')`), sin manejar el separador de miles colombiano (el punto). Combinado con
cómo el regex agrupa los números de cada línea del boletín, esto generó precios con un orden de magnitud
incorrecto en algunos productos — sobre todo `concentrado_aves`, `concentrado_cerdos`, `concentrado_gatos`,
`concentrado_perros`, `nucleo_proteico`, `harina_sangre` y `subproductos_carnicos`.

**Lo que se hizo en su momento:** se corrigió `a_float()` en el código del scraper, pero **no se reprocesaron
los PDFs** (se estimaron ~40 horas para los 878 boletines) — en su lugar, se filtró `bmc_precios_limpio` con
rangos de precio hardcoded para esconder los valores absurdos.

**El problema de esa solución:** el scraper guarda con `INSERT OR IGNORE` sobre `(fecha, producto_bmc)`. Eso
significa que los registros que ya estaban en la base **antes** del fix de `a_float()` nunca se sobrescriben
— siguen ahí, con el valor incorrecto, solo que ahora invisibles porque el filtro de `bmc_precios_limpio` los
descarta. El dato de fondo (`bmc_precios`, la tabla cruda) sigue dañado para esos productos.

Las celdas siguientes verifican esto con tus datos reales — para que la decisión de reprocesar (o no) se tome
con evidencia, no de memoria.

```python
# ── Diagnóstico: ¿el bug sigue presente en bmc_precios (tabla cruda)? ─────
PRODUCTOS_SENSIBLES_AL_BUG = [
    'concentrado_aves', 'concentrado_cerdos', 'concentrado_gatos', 'concentrado_perros',
    'nucleo_proteico', 'harina_sangre', 'subproductos_carnicos'
]

print("🔎 Filas con magnitud extrema (>50x la mediana del producto) en bmc_precios CRUDA:\n")
afectados = df_validez[df_validez['producto'].isin(PRODUCTOS_SENSIBLES_AL_BUG)]
display(afectados[['producto', 'n', 'mediana_cop_kg', 'extremos_50x_mediana', 'pct_extremos']])

total_filas_afectadas = afectados['extremos_50x_mediana'].sum()
print(f"\nTotal de filas con huella del bug: {total_filas_afectadas}")

if total_filas_afectadas > 0:
    print("\n🚨 CONCLUSIÓN: el bug SIGUE presente en los datos guardados.")
    print("   El fix en el código no corrigió retroactivamente lo que ya estaba en la DB")
    print("   (por el INSERT OR IGNORE). Toca decidir entre:")
    print("   a) Reprocesar TODOS los PDFs (~40h) con el a_float corregido, sobrescribiendo bmc_precios")
    print("   b) Reprocesar SOLO los PDFs que generaron filas afectadas (mucho más rápido)")
else:
    print("\n✅ CONCLUSIÓN: no se detecta huella del bug en los datos actuales.")
    print("   Es posible que ya se haya corregido en una corrida anterior.")
```

```python
# ── Si hay filas afectadas: ¿qué PDFs hay que reprocesar puntualmente? ────
if total_filas_afectadas > 0:
    pdfs_a_reprocesar = []
    for prod in PRODUCTOS_SENSIBLES_AL_BUG:
        fila_mediana = df_validez.loc[df_validez['producto'] == prod, 'mediana_cop_kg']
        if fila_mediana.empty:
            continue
        mediana = fila_mediana.iloc[0]
        umbral = mediana * 50
        sub = query(
            "SELECT DISTINCT archivo_pdf FROM bmc_precios WHERE producto = ? AND precio_cop_kg > ?",
            params=(prod, umbral)
        )
        pdfs_a_reprocesar.extend(sub['archivo_pdf'].tolist())

    pdfs_a_reprocesar = sorted(set(pdfs_a_reprocesar))
    print(f"📂 PDFs puntuales con al menos un valor extremo: {len(pdfs_a_reprocesar)}")
    print(f"   (de un total de 878 — reprocesar solo estos es mucho más rápido que el lote completo)")
    if pdfs_a_reprocesar:
        print("\nPrimeros 15:")
        for p in pdfs_a_reprocesar[:15]:
            print(f"   - {p}")
else:
    print("No aplica — no se detectaron filas afectadas.")
```

### Decisión a tomar

- Si **el número de PDFs afectados es pequeño** (decenas, no centenas): conviene **borrar esas filas
  puntuales de `bmc_precios`** (por `archivo_pdf`) y reprocesar solo esos PDFs con el `a_float()` corregido.
  Minutos, no horas.
- Si **es la mayoría de los 878**: ahí sí se justifica el reprocesamiento completo nocturno.
- En cualquier caso, **no basta con seguir filtrando `bmc_precios_limpio`** — eso esconde el problema para el
  modelo de precio de HCH, pero si en algún momento quieres usar `nucleo_proteico`, `harina_sangre` o
  `subproductos_carnicos` como variable explicativa (tienen alta correlación esperada con HCH), vas a estar
  modelando con una serie que tiene huecos de confianza no documentados.

```python
import sqlite3, pandas as pd
con = sqlite3.connect("mercado_hch.db")

PRODUCTOS_SENSIBLES_AL_BUG = [
    'concentrado_aves', 'concentrado_cerdos', 'concentrado_gatos', 'concentrado_perros',
    'nucleo_proteico', 'harina_sangre', 'subproductos_carnicos'
]

df = pd.read_sql("SELECT producto, precio_cop_kg FROM bmc_precios", con)
medianas = df.groupby('producto')['precio_cop_kg'].median()

for prod in PRODUCTOS_SENSIBLES_AL_BUG:
    if prod not in medianas.index:
        continue
    mediana = medianas[prod]
    umbral = mediana * 50
    n_extremos = (df[(df['producto']==prod) & (df['precio_cop_kg'] > umbral)]).shape[0]
    print(f"{prod}: {n_extremos} filas con valor extremo")

con.close()
```

```python
con = sqlite3.connect("mercado_hch.db")

mediana = medianas['subproductos_carnicos']
umbral = mediana * 50

df_afectados = pd.read_sql(f"""
    SELECT archivo_pdf, fecha, precio_cop_kg
    FROM bmc_precios
    WHERE producto = 'subproductos_carnicos' AND precio_cop_kg > {umbral}
""", con)

print(df_afectados.to_string(index=False))
con.close()
```

```python
con = sqlite3.connect("mercado_hch.db")

df_detalle = pd.read_sql(f"""
    SELECT archivo_pdf, fecha, cantidad_kg, total_cop, precio_cop_kg,
           ROUND(total_cop / cantidad_kg, 2) AS precio_recalculado
    FROM bmc_precios
    WHERE producto = 'subproductos_carnicos' AND precio_cop_kg > {umbral}
""", con)

print(df_detalle.to_string(index=False))
con.close()
```
